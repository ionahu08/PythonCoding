```python
import numpy as np
import pandas as pd
```

```python
X = np.random.randn(1000, 3)
bias = np.ones((1000, 1))
X_b = np.hstack((bias, X))
true_weights = np.array([[2.5],[3.0],[4.0],[5.0]])
y = X_b @ true_weights + 0.1*np.random.randn(1000, 1)
```

```python
def multi_linear_regression(X):
    theta = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
    return theta
theta_best = multi_linear_regression(X_b)
```

```python
def pred(X, theta_best):
    preds = X @ theta_best
    return preds
preds = pred(X_b, theta_best)
```

```python
def RMSELoss(pred, actual):
    return np.sqrt(np.mean((pred - actual)**2))

RMSELoss(preds, X_b)
```

```python
np.hstack((preds, y))
```

```python
X = np.random.randn(1000, 4)
bias = np.full((1000, 1), 4)
X_b = np.hstack((bias, X))
true_weights = np.array([[3.0],[4.0],[2.0], [1.0],[-9.0]])
y = X_b @ true_weights + np.random.randn(1000, 1)*0.1
```

```python
def LRModel(X, y):
    best_theta = np.linalg.inv(X.T @ X) @ X.T @ y
    return best_theta

best_theta = LRModel(X_b, y)
```

```python
def pred(X, best_theta):
    res = X @ best_theta
    return res
preds = pred(X_b, best_theta)
```

```python
def RSMLoss(pred, actual):
    RSMLoss = np.sqrt(np.mean((pred - actual)**2))
    return RSMLoss
RSMLoss(preds, y)
```
