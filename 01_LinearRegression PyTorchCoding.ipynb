```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
```

```python
# Create a sample data, including 100 samples and 3 features
X = torch.randn(100, 3)
true_weights = torch.tensor([[2.0],[-1.0],[3.5]])
true_bias = 4.0
y = X @ true_weights + true_bias + 0.1*torch.randn(100, 1) # add small noises
```

```python
dataset = TensorDataset(X, y)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)
```

```python
# Define a simple linear model
class LinearRegressionModel(nn.Module):  # Define a LinearRegressionModel class that inherits from nn.Module
    def __init__(self, input_dim):
        super().__init__()
        self.linear = nn.Linear(input_dim, 1)

    def forward(self, x):  # forward function defines how the model compute its output
        return self.linear(x)
```

```python
model = LinearRegressionModel(input_dim = 3)
```

######  An optimizer is an algorithm that updates models `weights` to minimize `loss` during Training
######  SGD vs Adam
######    - SGD: Stochastic Gradient Descent, simple and widely used
######    - Adam: Adaptive moment estimation, combines the advantages of SGD with momentum and RMSProp, often performs better out of the box.

```python
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)     
```

```python
for epoch in range(500):
    for batch_X, batch_y in dataloader:
        model.train()  # training model
    
        # forward
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
    
        # backward 
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()


    #evaluation
    model.eval()
    preds = model(batch_X)
    eval_loss = criterion(preds, batch_y)

    if epoch % 100 == 0:
        print(f"Epoch:{epoch}, Training loss: {loss.item():.4f}, Eval Loss: {eval_loss.item():.4f}")
        
```

```python
print("model weights data:", model.linear.weight.data)
print("model bias data:", model.linear.bias.data)
```
