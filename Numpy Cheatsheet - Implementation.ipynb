```python
import numpy as np
```

### 01. Creating Arrays

```python
np.array([1,2,3])                         # 1D array
np.array([[1,2,3],[3,4,5]])               # 2D array

np.zeros((2,3))                           # Array of zeros
np.ones((2,3))                            # Array of zeros
np.full((2,3), 10)                        # Array of constant

np.arange(0, 12, 2)                       # Equal spaced values
np.linspace(0, 12, 4)                     # 4 values from 0 to 12
np.eye(4)                                 # Identity Matrix
np.random.rand(2,3)                       # Create a 2*3 Matrix of uniformly distributed random numbers
np.random.randn(2,3)                      # Create a 2*3 Matrix of normally distributed random numbers
```

### 02.Array Properties

```python
arr = np.array([[1,2,"3"],[4,5,"6"]])
```

```python
arr.shape      # Dimension
arr.size       # Number of Elements
arr.ndim       # Number of Dimensions
arr.dtype      # Data type of Elements
```

### 03. Array Operations
##### 01 - Element-wise operations
##### 02 - Matrix operations

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[2,4,6],[8,10,12]])
c = np.array([[1,1],[2,2]])
d = np.array([1,4,9,16])
```

```python
# 01 - Element-wise operations
a*b
a+b
a-b
a/b
np.sqrt(d)
np.exp(c)
np.log(c)
```

```python
# 02 - Matrix operations
a.T                # Transpose
np.dot(a, a.T)     # Dot product
np.matmul(a,a.T)   # Note: when applied to 2D Matrix, it's equvalent to np.dot; but it doesn't support operations between 1D Matrix 
```

### 04. Indexing & Slicing

```python
a[0,1]           # Select one cell
a[:, 1]          # Select one column
a[0, :]          # Select one row
a[0:2, 1:2]      # Select a Subarray
a[a>4]           # Bool
```

### 05. Manipulation Arrays (Reshaping)

```python
np.reshape(a, (3,2))        # change shape
np.ravel(a)                 # Flatten Array
np.concat([a,b], axis=0)    # Append vertically
np.concat([a,b], axis=1)    # Append horizontally
np.vstack((a,b))            # Vertical stck
np.hstack((a,b))            # Horizontal stck
```

### 06. Aggregation & Statistics

```python
a.sum()
a.std()
a.var()
a.min(axis=0)   # min Across rows
a.max()
a.mean()

a.argmin()     # Returns the index of the min/max value in a NumPy array.
a.argmax()     # Returns the index of the min/max value in a NumPy array.
```

### 07.Cleaning & NaN Handling

```python
a = a.astype(float)
a[0, 1] = np.nan
np.isnan(a).sum()   # count missing
```

### 08. Utilities

```python
np.unique(a)
np.sort(a)
np.where(a>5, "larget","small")
np.clip(a, 0, 3)    #limit the values in a within the interval [0,3]
```
